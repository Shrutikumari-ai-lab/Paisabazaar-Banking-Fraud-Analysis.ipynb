# **Project Summary -**

This project focuses on analyzing customer financial data to understand the factors that influence credit scores. Credit score is an important measure used by banks and financial institutions to evaluate the creditworthiness of a person. It helps lenders decide whether a person is eligible for loans, credit cards, or other financial products. A good credit score shows that a person is financially responsible and pays their debts on time, while a poor credit score indicates a higher risk for lenders.

Paisabazaar is a financial services platform that helps customers compare and apply for different financial products such as personal loans, credit cards, and insurance. For companies like Paisabazaar, understanding the credit behavior of customers is very important because it helps reduce the risk of loan defaults and improves financial decision-making. In this project, Exploratory Data Analysis (EDA) is used to study customer data and identify the patterns and relationships that affect credit scores.

The dataset used in this project contains several attributes related to customers and their financial activities. These attributes include information such as age, annual income, number of bank accounts, number of credit cards, outstanding debt, payment behavior, number of loans, and credit utilization ratio. Each of these factors may influence a person’s credit score. By analyzing these variables, we can better understand which financial behaviors are associated with good or poor credit scores.

The first step in the project is data loading and data cleaning. The dataset is imported using Pandas, which is commonly used for data manipulation and analysis in Python. During the cleaning process, missing values, duplicate records, and incorrect data types are handled to ensure that the dataset is accurate and reliable for analysis. Cleaning the data is an important step because it improves the quality of the results obtained from the analysis.

After the data is cleaned, Exploratory Data Analysis (EDA) is performed to understand the structure and distribution of the dataset. EDA helps in discovering patterns, trends, and relationships between different variables. Various statistical summaries and visualizations are used to examine the data. For example, the distribution of credit score categories can be analyzed to see how many customers fall into the Good, Standard, and Poor categories.

Data visualization plays a key role in this project. Graphs and charts make it easier to understand complex datasets and communicate insights clearly. Visualization libraries such as Matplotlib, Seaborn, and Plotly are used to create bar charts, histograms, and other visual representations of the data. These visualizations help in identifying relationships such as how income affects credit scores or how outstanding debt is related to financial behavior.

Through the analysis, several insights can be obtained. For example, customers with higher income may have better credit scores, while customers with higher outstanding debt may be more likely to have lower credit scores. Payment behavior and credit utilization also play an important role in determining creditworthiness. These observations help in understanding how different financial factors influence credit score categories.

Overall, the purpose of this project is to explore and understand the dataset using Exploratory Data Analysis techniques. The findings from this analysis can help financial service companies make better decisions when evaluating customers and offering financial products. The project also demonstrates how data analysis can be used to gain meaningful insights from financial datasets and support better decision-making in the financial services industry.

# KPIs(Key Perfromance Indicators) Calculation

**1. Customer Demographics**

````python
#1 - Customer Demographics

#Average Age of Customers
avg_age = df1['Age'].mean()
print(f"Average Age of Customers: {avg_age:.2f} years")
````

````python
#Distribution of Occupation
occupation_counts = df1['Occupation'].value_counts()
print("Distribution of Occupation:")
print(occupation_counts)
````

````python
#Number of Customers by Income Group

# Define income groups (bins)
max_income = df['Annual_Income'].max()
bins = [0, 25000, 50000, 100000, 200000]

labels = ['Low (<25K)', 'Lower-Middle (25K-50K)',
          'Middle (50K-100K)', 'Upper-Middle (100K-200K)']

if max_income > 200000:
    bins.append(max_income)
    labels.append('High (>200K)')

# Create income group column
df['Income_Group'] = pd.cut(df['Annual_Income'], bins=bins, labels=labels, include_lowest=True)

# KPI: Number of customers by income group
income_group_counts = df['Income_Group'].value_counts().sort_index()

print(income_group_counts)
````

**2- Financial Behavior**

````python
#Average Annual Income
avg_annual_income = df['Annual_Income'].mean()
print(f"Average Annual Income: ${avg_annual_income:.2f}")
````

````python

#Average Monthly In-hand Salary
avg_monthly_salary = df['Monthly_Inhand_Salary'].mean()
print(f"Average Monthly In-hand Salary: ${avg_monthly_salary:.2f}")
````

````python
#Average Debt-to-Income Ratio
debt_to_income_ratio = df['Outstanding_Debt'] / df['Annual_Income']
avg_debt_to_income_ratio = debt_to_income_ratio.mean()
print(f"Average Debt-to-Income Ratio: {avg_debt_to_income_ratio:.2f}")
````

````python

#Average Credit Utilization Ratio
credit_utilization_ratio = df['Credit_Utilization_Ratio']
avg_credit_utilization_ratio = credit_utilization_ratio.mean()
print(f"Average Credit Utilization Ratio: {avg_credit_utilization_ratio:.2f}")
````

````python

#Average Number of Bank Accounts
avg_num_bank_accounts = df['Num_Bank_Accounts'].mean()
print(f"Average Number of Bank Accounts: {avg_num_bank_accounts:.2f}")
````
